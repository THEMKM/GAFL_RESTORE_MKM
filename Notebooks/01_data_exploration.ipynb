```python
import pandas as pd

# Define file path
file_path = "../data/raw/cdi_household_clean_nopii.dta"  # Adjust path as needed

# Load the .dta file
df = pd.read_stata(file_path, convert_categoricals=False)

# Display basic info
df.info()
df.head()
```

```python
# Count missing values per column
df.isnull().sum().sort_values(ascending=False)
```

```python
df.describe()
```

```python
for col in df.columns:
    if df[col].dtype == 'object':  # Text-based columns
        print(f"{col}: {df[col].nunique()} unique values")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
sns.histplot(df["cocoa_yield"], bins=30, kde=True)
plt.title("Distribution of Cocoa Yield (kg/ha)")
plt.show()
```
