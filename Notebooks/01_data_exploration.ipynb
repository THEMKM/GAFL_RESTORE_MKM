```python
import pandas as pd

# Define the path to your Stata file
data_path = "../data/raw/cdi_household_clean_nopii.dta"
df = pd.read_stata(data_path, convert_categoricals=False)

# Programmatically build the list of cocoa_ages columns for farms 1 to 5
# If you also have 'cocoa_ages_tot_X', add "tot" to age_groups
age_groups = ["0_5", "6_10", "11_19", "20_30", "31", "tot"]
cocoa_ages_clones = [f"cocoa_ages_{age}_{i}" for age in age_groups for i in range(1, 6)]

# Now define all other columns you want to select
other_columns = [
    # Cocoa production variables
    "cocoa_produce_main", "cocoa_produce_light", "landuseha",
    # Fertilizer and pesticide variables
    "fert_cons", "fert_size", "fert_cost", "fert_lqd_cons", "fert_lqd_cost",
    "herbicide_num_app", "herbicide_cost", "pesticide_num_app", "pesticide_cost",
    # Labor cost variable
    "total_labor_cost",
    # Price variables
    "cocoa_price", "cocoa_price_light",
    # Agroforestry variables
    "shade_planted", "shade_tree_tot",
    # Tenure and dispute variables
    "doc_land", "encroach_family", "encroach_vol", "encroach_chief",
    "encroach_landlord", "encroach_timber", "encroach_frstdpt",
    "farm_yracquired", "conflict_land_num",
    # Mining variables
    "mining_intensity", "mining_distance", "income_mining_amt",
    # Income variables
    "income_cocoa", "income_noncocoa", "income_pension",
    # Household characteristics
    "hhmem_size", "hh_education_level", "access_extension",
    "training_participation", "access_credit",
]

# Combine the cocoa_ages clones with the other columns
selected_columns = cocoa_ages_clones + other_columns

# If you want to avoid errors in case some columns don't exist, you can filter:
selected_columns = [col for col in selected_columns if col in df.columns]

# Create the new DataFrame
df_selected = df[selected_columns].copy()

print("Selected features shape:", df_selected.shape)
print("Selected columns:", df_selected.columns.tolist())
```

```python
# Compute the weighted average age of cocoa trees
df_selected['avg_cocoa_age'] = (
    df_selected['cocoa_ages_0_5'] * 2.5 +
    df_selected['cocoa_ages_6_10'] * 8 +
    df_selected['cocoa_ages_11_19'] * 15 +
    df_selected['cocoa_ages_20_30'] * 25 +
    df_selected['cocoa_ages_31'] * 35.5
) / 100

print("Computed 'avg_cocoa_age'.")
```

```python
# Calculate cocoa yield per hectare (kg/ha)
df_selected['cocoa_yield_ha'] = (
    df_selected['cocoa_produce_main'] + df_selected['cocoa_produce_light']
) / df_selected['landuseha']

print("Computed 'cocoa_yield_ha'.")
```

```python
# Calculate fertilizer cost per hectare (CFA/ha)
df_selected['fert_cost_ha'] = (
    (df_selected['fert_cons'] * df_selected['fert_size'] * df_selected['fert_cost']) +
    (df_selected['fert_lqd_cons'] * df_selected['fert_lqd_cost'])
) / df_selected['landuseha']

print("Computed 'fert_cost_ha'.")
```
