```python
# %% Step 2: Enhanced Search Algorithm (Fixed)
import pandas as pd
from fuzzywuzzy import process

def find_matches(var_name, existing_vars, threshold=70):
    """Find matches with proper index handling"""
    # Normalize names
    normalized_var = var_name.lower().replace('_', '')
    normalized_dta = [v.lower().replace('_', '') for v in existing_vars]
    
    # Get matches (returns tuples of (normalized_dta_string, score))
    matches = process.extract(normalized_var, normalized_dta, limit=30)
    
    potential_matches = []
    for match_nv, score in matches:
        if score >= threshold:
            # Find original variable using the index
            idx = normalized_dta.index(match_nv)
            potential_matches.append(existing_vars[idx])
    
    return list(set(potential_matches))  # Remove duplicates

# Load data
df = pd.read_stata("../data/raw/cdi_household_clean_nopii.dta", convert_categoricals=False) 
existing_vars = df.columns.tolist()

# Create matching dataframe
matches = []
for idx, var in enumerate(pdf_variables, 1):
    if var in existing_vars:
        matches.append({'counter': idx, 'pdf_var': var, 'dta_var': var, 'match_type': 'exact'})
    else:
        potential_matches = find_matches(var, existing_vars)
        if potential_matches:
            matches.append({
                'counter': idx,
                'pdf_var': var,
                'dta_var': " | ".join(potential_matches),
                'match_type': 'fuzzy'
            })
        else:
            matches.append({
                'counter': idx,
                'pdf_var': var,
                'dta_var': "NOT FOUND",
                'match_type': 'none'
            })

matches_df = pd.DataFrame(matches)
```

```python
# Save interactive CSV for manual selection
matches_df['selected_dta_var'] = matches_df['dta_var']
matches_df.to_csv("../data/processed/variable_mapping_template.csv", index=False)

print("""
1. Open 'variable_mapping_template.csv' in Excel/Sheets
2. For rows with multiple matches (separated by |):
   - Replace the value in 'selected_dta_var' with the correct variable name
   - Delete unwanted options
3. Save and use the cleaned file for analysis
""")
```
